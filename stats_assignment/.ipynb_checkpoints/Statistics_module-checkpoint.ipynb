```python
import statistics as stats
import numpy as np
import pandas as pd
import matplotlib as plt
```

```python
X = [2,4,6,8,3,5,1,10,12,11]
Y = [4,5,6,7,8,9,10,12,13,11]
```

```python
print("X:",X)
print("Y:",Y)
print("-" *50)
```

```python

plt.hist(X)
plt.xlabel("X values")
plt.ylabel("Frequency")
plt.title("Histogram of X")
plt.show()
```

```python

plt.hist(Y)
plt.xlabel("Y values")
plt.ylabel("Frequency")
plt.title("Histogram of Y")
plt.show()
```

```python
mode_x = stats.mode(X)
mode_y = stats.mode(Y)
print(mode_x)
print(mode_y)
```

```python
plt.boxplot(X)
plt.ylabel("X values")
plt.title("Box Plot of X")
plt.show()
```

```python
plt.boxplot(Y)
plt.ylabel("Y values")
plt.title("Box Plot of Y")
plt.show()
```
