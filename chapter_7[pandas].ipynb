### Recognizing missing data

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, np.nan, 30],         # Bob's Age is missing
    'Salary': [50000, 60000, np.nan] # Charlie's Salary is missing
})

# 1. The Mask (Hard to read for big data)
print(df.isnull()) 
```

```python
# 2. The Count (The Standard Way)
print("Missing values per column:")
print(df.isnull().sum())
```

#### Handling Missing Data

#### Removing NaN values

```python
import pandas as pd

# Original DataFrame
print("Original DataFrame:")
print(df)
print("-" * 50)

# Remove ANY row that contains at least one NaN
clean_rows = df.dropna()
print("After dropping rows with ANY NaN:")
print(clean_rows)
print("-" * 50)

# Remove ANY column that contains at least one NaN
clean_cols = df.dropna(axis=1)
print("After dropping columns with ANY NaN:")
print(clean_cols)
print("-" * 50)

# Smart drop: Remove rows ONLY if 'Name' is NaN
# (Keeps rows where other columns like Age may be NaN)
smart_drop = df.dropna(subset=["Name"])
print("After dropping rows where 'Name' is NaN:")
print(smart_drop)
print("-" * 50)
```

```python
# 1. Fill with Zero (Good for things like 'Debt' or 'Points')
df_zero = df.fillna(0)
print("After filling NaN values with 0 ")
print(df_zero)
print("-" * 50)
```

#### Statistical Imputation

```python
print("Before filling NaN in Age:")
print(df)
print("-" * 50)

mean_age = df["Age"].mean()
print(f"Mean Age: {mean_age}")
print("-" * 50)

# Fill NaN values in Age with the mean
df["Age"] = df["Age"].fillna(mean_age)

# After filling NaN
print("After filling NaN in Age with mean:")
print(df)
print("-" * 50)
```

### Handling Duplicates

```python
import pandas as pd

# Sample Dataset with Duplicates
data = {
    'TransactionID': [101, 102, 101, 103],
    'Amount': [500, 750, 500, 200]
}
df = pd.DataFrame(data)
print(df)
print("-" * 50)
# Step 1: Identification (Returns Boolean Truth Values)
# True = This row is a duplicate of a previous row
print(f'Which rows are duplicates?')
print(df.duplicated()) 

# Step 2: Quantification (Counting Errors)
# Summing the boolean values (True=1, False=0)
duplicate_count = df.duplicated().sum()
print(f"Total Duplicate Rows: {duplicate_count}")
print("-" * 50)

# Step 3: Cleaning (The 'Set' Operation)
# 'inplace=True' modifies the actual table instead of creating a copy
df.drop_duplicates(inplace=True)

# Verification
print(df)
```

### Detecting Outliers

$$Z = \frac{X - \mu}{\sigma}$$

```python
import pandas as pd

print("STEP 1: Create sample salary data")
print("-" * 50)

df = pd.DataFrame({'Salary': [48, 50, 52, 49, 51, 53, 47, 34, 41, 22, 93, 72, 10, 45, 88, 1, 170]})

print("Original Data:")
print(df)
print("\n")
```

```python
print("STEP 2: Compute statistical properties")
print("-" * 50)

# Mean (μ): central tendency
mean_salary = df['Salary'].mean()

# Standard deviation (σ): spread of the data
std_dev = df['Salary'].std()

print(f"Mean (μ) Salary      : {mean_salary:.2f}")
print(f"Std Deviation (σ)    : {std_dev:.2f}")
print("\n")
```

```python
print("STEP 3: Define the Safe Zone using the 3-Sigma Rule")
print("-" * 50)

upper_limit = mean_salary + (3 * std_dev)
lower_limit = mean_salary - (3 * std_dev)

print(f"Lower Limit (μ - 3σ): {lower_limit:.2f}")
print(f"Upper Limit (μ + 3σ): {upper_limit:.2f}")
print("\n")
```

```python
print("STEP 4: Remove values outside the Safe Zone (Outliers)")
print("-" * 50)

# Keep only values within the defined boundaries
df_clean = df[
    (df['Salary'] > lower_limit) |
    (df['Salary'] < upper_limit)
]

# df_lower=df[
#     (df['Salary']>lower_limit)
# ]

# print(df_lower)
print("Cleaned Data using or (Outliers Removed):")
print(df_clean)


df_clean = df[
    (df['Salary'] > lower_limit) &
    (df['Salary'] < upper_limit)
]

# df_lower=df[
#     (df['Salary']>lower_limit)
# ]

# print(df_lower)
print("Cleaned Data using and  (Outliers Removed):")
print(df_clean)
```

```python
#### Extreme Outliers
```

```python
import pandas as pd

df = pd.DataFrame({
    'Salary': [48, 50, 52, 49, 51, 53, 10_000_000]
})
df
```

```python
mu = df['Salary'].mean()
sigma = df['Salary'].std()
print(f'Mean: {mu}, Standard Deviation: {sigma}')
print("-" * 50)

lower = mu - 3*sigma
upper = mu + 3*sigma

print(f"3σ Range: {lower:.2f} to {upper:.2f}")
print("-" * 50)

print(df[(df['Salary'] > lower) & (df['Salary'] < upper)])
```

Here, 3-sigma rule fails. 
Extreme value remains.

Thus, we need inter-quartile range for better handling of such extreme outliers.

#### Inter-quartile Range


```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

print(f"IQR Range: {lower:.2f} to {upper:.2f}")
print(df[(df['Salary'] >= lower) & (df['Salary'] <= upper)])
```

### Data Transformation, Normalization and Type Conversion

```python
import pandas as pd

# ----------------------------
# Sample dataset (raw data)
# ----------------------------
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male'],   # Nominal categorical
    'Risk': ['Low', 'High', 'Medium'],     # Ordinal categorical
    'Age': ['25', '45', '30']              # Numeric stored as strings
})

print("Original Data:")
print(df)
print("\n")

# ----------------------------
# 1. Categorical Encoding (Nominal)
# ----------------------------
# Gender is nominal: no order
# Mapping text to integers using dictionary (Unit V: Key-Value)
gender_map = {'Male': 0, 'Female': 1}
df['Gender_encoded'] = df['Gender'].map(gender_map)

print("After Categorical Encoding (Nominal - Gender):")
print(df[['Gender', 'Gender_encoded']])
print("\n")

# ----------------------------
# 2. Ordinal Encoding
# ----------------------------
# Risk is ordinal: Low < Medium < High
# Mapping preserves order
risk_map = {'Low': 0, 'Medium': 1, 'High': 2}
df['Risk_encoded'] = df['Risk'].map(risk_map)

print("After Ordinal Encoding (Risk):")
print(df[['Risk', 'Risk_encoded']])
print("\n")
```

```python

# ----------------------------
# 3. Type Conversion
# ----------------------------
# Age is numeric but stored as string
# Convert to integer for computation
df['Age'] = df['Age'].astype(int)

print("After Type Conversion (Age):")
print(df[['Age']])
print("\n")

# ----------------------------
# 4. Normalization (Min-Max Scaling)
# ----------------------------
# Scale Age to [0,1] range
df['Age_norm'] = (df['Age'] - df['Age'].min()) / (df['Age'].max() - df['Age'].min())

print("After Normalization (Age):")
print(df[['Age', 'Age_norm']])
print("\n")

# ----------------------------
# Final DataFrame
# ----------------------------
print("Final DataFrame with all transformations applied:")
print(df)
```
